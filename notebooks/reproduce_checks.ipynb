# Reproduce the Clay v1.5 paper checks

This notebook runs the repository's CPU checks from a clean clone. It needs no GPU and no account.

1. `make env`: installs the pinned environment from `uv.lock` (about 2 GB).

3. `make smoke`: downloads the Clay v1.5 checkpoint (5.2 GB), verifies its SHA-256, loads it strictly, and reproduces the published teacher-alignment cosines on one Sentinel-2 scene.

Expect about 20 to 40 minutes on a free Colab CPU runtime, mostly downloads. The GPU experiments (Table 2 benchmark runs, E1 to E8) are listed in `REPRODUCING.md`.
